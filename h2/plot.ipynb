```python
import matplotlib.ticker as mtick
import matplotlib.pyplot as plt
import numpy as np
import h5py
#from common import smooth
from matplotlib.ticker import FormatStrFormatter

plt.rc('font', family='Times New Roman', size=14)
plt.rcParams['mathtext.fontset'] = 'custom'

plt.rcParams['mathtext.it'] = 'STIXGeneral:italic'
plt.rcParams['mathtext.rm'] = 'STIXGeneral:bold'
plt.rcParams['mathtext.tt'] = 'STIXGeneral'
plt.rcParams['mathtext.bf'] = 'STIXGeneral:bold:italic'
plt.rcParams['mathtext.cal'] = 'STIXGeneral'
plt.rcParams['mathtext.sf'] = 'STIXGeneral'

plt.rcParams['lines.linewidth'] = 1.5
plt.rcParams['axes.linewidth'] = 1.5

itr = r'$R$'
ite = r'$E$'

color_dict = {'HF':'tab:gray',
              'CASSCF':'tab:gray',
              'GW':'tab:red',
              'GW1':'tab:blue',
              'GW2':'tab:red',
              'FCI':'black'}

color_dict = {'HF':'C1',
              'CASSCF':'C0',
              'GW0':'tab:green',
              'GW1':'tab:olive',
              'GW2':'tab:red',
              'FCI':'black'}
```

```python
from gf.gw.addons import read_spec_dat

from pyscf.data.nist import HARTREE2EV as au2ev
```

```python
rr = 1.

W, dat_rhfgw_1 = read_spec_dat(f'spec-GW@RHF-{rr:2.1f}.dat')
W, dat_cas_1 = read_spec_dat(f'spec-CAS-{rr:2.1f}.dat')
W, dat_casgw_1 = read_spec_dat(f'spec-GW@CAS-{rr:2.1f}.dat')
W, dat_fci_1 = read_spec_dat(f'spec-FCI-{rr:2.1f}.dat')
W, dat_uhfgw_1 = read_spec_dat(f'spec-GW@UHF-{rr:2.1f}.dat')
```

```python
rr = 2.

W, dat_rhfgw_2 = read_spec_dat(f'spec-GW@RHF-{rr:2.1f}.dat')
W, dat_cas_2 = read_spec_dat(f'spec-CAS-{rr:2.1f}.dat')
W, dat_casgw_2 = read_spec_dat(f'spec-GW@CAS-{rr:2.1f}.dat')
W, dat_fci_2 = read_spec_dat(f'spec-FCI-{rr:2.1f}.dat')
W, dat_uhfgw_2 = read_spec_dat(f'spec-GW@UHF-{rr:2.1f}.dat')
```

```python
rr = 3.

W, dat_rhfgw_3 = read_spec_dat(f'spec-GW@RHF-{rr:2.1f}.dat')
W, dat_cas_3 = read_spec_dat(f'spec-CAS-{rr:2.1f}.dat')
W, dat_casgw_3 = read_spec_dat(f'spec-GW@CAS-{rr:2.1f}.dat')
W, dat_fci_3 = read_spec_dat(f'spec-FCI-{rr:2.1f}.dat')
W, dat_uhfgw_3 = read_spec_dat(f'spec-GW@UHF-{rr:2.1f}.dat')
```

```python
fig,(ax1,ax2,ax3) = plt.subplots(nrows=3,ncols=1,figsize=(8,6))

ax1.set_xlim(-.8*au2ev,.28*au2ev)
ax2.set_xlim(-.8*au2ev,.28*au2ev)
ax3.set_xlim(-.8*au2ev,.28*au2ev)

ax1.set_ylim(0,200/au2ev)
ax2.set_ylim(0,200/au2ev)
ax3.set_ylim(0,200/au2ev)

#ax1s = fig.add_axes((.55,.72,.15,.1))

ax1.plot(W*au2ev, dat_rhfgw_1/au2ev, label=r'$GW$@RHF', color=color_dict['GW0'])
ax1.plot(W*au2ev, dat_uhfgw_1/au2ev, label=r'$GW$@UHF', color=color_dict['GW1'])
ax1.plot(W*au2ev, dat_cas_1/au2ev, label='CAS(2,2)', color=color_dict['CASSCF'])
ax1.plot(W*au2ev, dat_casgw_1/au2ev, label=r'MR-$GW$(2,2)', color=color_dict['GW2'])
ax1.plot(W*au2ev, dat_fci_1/au2ev, label='FCI', color=color_dict['FCI'], ls='--')

ax2.plot(W*au2ev, dat_rhfgw_2/au2ev, label=r'$GW$@RHF', color=color_dict['GW0'])
ax2.plot(W*au2ev, dat_uhfgw_2/au2ev, label=r'$GW$@UHF', color=color_dict['GW1'])
ax2.plot(W*au2ev, dat_cas_2/au2ev, label='CAS(2,2)', color=color_dict['CASSCF'])
ax2.plot(W*au2ev, dat_casgw_2/au2ev, label=r'$GW$@CAS', color=color_dict['GW2'])
ax2.plot(W*au2ev, dat_fci_2/au2ev, label='FCI', color=color_dict['FCI'], ls='--')

ax3.plot(W*au2ev, dat_rhfgw_3/au2ev, label=r'$GW$@RHF', color=color_dict['GW0'])
ax3.plot(W*au2ev, dat_uhfgw_3/au2ev, label=r'$GW$@UHF', color=color_dict['GW1'])
ax3.plot(W*au2ev, dat_cas_3/au2ev, label='CAS(2,2)', color=color_dict['CASSCF'])
ax3.plot(W*au2ev, dat_casgw_3/au2ev, label=r'$GW$@CAS', color=color_dict['GW2'])
ax3.plot(W*au2ev, dat_fci_3/au2ev, label='FCI', color=color_dict['FCI'], ls='--')

#ax1s.plot(W*au2ev, dat_cas_1, label='CAS(2,2)', color=color_dict['CASSCF'])
#ax1s.set_xlim(left=.6*au2ev, right=.75*au2ev)

#ax1s.set_xticks([17,18,19,20])
#ax1s.set_yticks([0,50,100,150])
#ax1s.set_xticklabels(labels=[17,18,19,20], fontsize=10)
#ax1s.set_yticklabels(labels=[0,50,100,150], fontsize=10)

ax3.set_xlabel(r'$\omega$/eV')
ax1.set_ylabel(r'$A(\omega)$')
ax2.set_ylabel(r'$A(\omega)$')
ax3.set_ylabel(r'$A(\omega)$')

ax1.set_xticklabels([])
ax2.set_xticklabels([])

ax1.legend(loc='center', bbox_to_anchor=(.5,1.2), ncol=5, fontsize=12, columnspacing=.6)

ax1.text(-8,160/au2ev, r'$R/R_0 = 1$')
ax2.text(-8,160/au2ev, r'$R/R_0 = 2$')
ax3.text(-8,160/au2ev, r'$R/R_0 = 3$')

fig.savefig(fname='spec-h2-123.pdf', bbox_inches='tight', format='pdf')
```
